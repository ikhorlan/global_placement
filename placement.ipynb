Linear regression

```python
df['tier_num'] = df['college_tier'].map({'Tier 1': 1, 'Tier 2' : 2, 'Tier 3': 3})
```

```python
df.head()
```

```python
df.info()
```

```python
df['rank_num']=df['university_ranking_band'].map({'Top 100' : 1, '100-300' : 2, '300+':3})
```

```python
df['placement_yes'] = df['placement_status'].map({'Placed': 1, 'Not Placed': 2})
```

```python
df.head()
```

```python
import seaborn as sns
```

```python
df_num = (
    df
    .select_dtypes(include='number')
    .dropna(axis = 'columns')
)
```

```python
df_num.info()
```

```python
X = df_num.drop(columns = ['salary', 'placement_yes'])
y = df_num.salary
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.score(X_test, y_test)
```

```python
X.columns
```

```python
print(lr.coef_, lr.intercept_)
```

```python
coef_table = (pd.DataFrame({'variable': X.columns, 'coefficient':lr.coef_}).round(2).sort_values('coefficient', key=abs, ascending=False).reset_index(drop=True))
```

```python
coef_table
```
